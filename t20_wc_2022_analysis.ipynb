```python
# -*- coding: utf-8 -*-
"""T20 World Cup 2022 - Comprehensive Analysis & Machine Learning Pipeline"""

import os
import re
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# ML Libraries
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
from sklearn.ensemble import RandomForestClassifier, RandomForestRegressor
from sklearn.linear_model import LogisticRegression, Ridge
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix, mean_absolute_error, r2_score

# Styling
plt.style.use('seaborn-v0_8-whitegrid' if 'seaborn-v0_8-whitegrid' in plt.style.available else 'default')
sns.set_palette("tab10")

# ==========================================
# 1. ROBUST DATA LOADING & INSPECTION
# ==========================================
# Check current directory or Colab default path
file_path = 'T20-world-cup-22.csv'
if not os.path.exists(file_path):
    file_path = '/content/sample_data/T20-world-cup-22.csv'

df = pd.read_csv(file_path)

# Standardize column names (lowercase & stripped spaces for easier coding)
df.columns = df.columns.str.strip().str.lower()
print("Dataset Shape:", df.shape)
print("Columns:", list(df.columns))
print(df.head(3))

# ==========================================
# 2. DATA CLEANING & PREPROCESSING
# ==========================================
# Drop duplicates properly
df = df.drop_duplicates()

# Fill missing scores/wickets if any match had rain/DLS
if 'first innings score' in df.columns:
    df['first innings score'] = pd.to_numeric(df['first innings score'], errors='coerce')
if 'second innings score' in df.columns:
    df['second innings score'] = pd.to_numeric(df['second innings score'], errors='coerce')

# Drop any matches with abandoned results / no winner
df = df.dropna(subset=['winner', 'first innings score'])

# ==========================================
# 3. FEATURE ENGINEERING & METRIC CORRECTIONS
# ==========================================
# A. Corrected Innings Run Rate (formerly 'economy rate')
df['first_innings_run_rate'] = df['first innings score'] / 20.0

# B. Top Scorer's Percentage Contribution (formerly labeled as strike rate)
# Strike rate requires balls faced; this measures reliance on a single player
df['top_scorer_pct_contribution'] = (df['highest score'] / df['first innings score']) * 100

# C. Match outcome characteristics
# Check if the team that won the toss also won the match
df['toss_match_win_match'] = (df['toss winner'] == df['winner'])

# D. Parse 'won by' into numerical margin & margin type (Runs vs Wickets)
def parse_margin(val):
    if pd.isna(val):
        return np.nan, 'Unknown'
    val_str = str(val).lower()
    num_match = re.search(r'\d+', val_str)
    num = int(num_match.group()) if num_match else np.nan
    if 'run' in val_str:
        return num, 'Runs'
    elif 'wicket' in val_str:
        return num, 'Wickets'
    return num, 'Other'

margin_info = df['won by'].apply(parse_margin)
df['margin_value'] = [m[0] for m in margin_info]
df['margin_type'] = [m[1] for m in margin_info]

# E. Did Batting 1st or Chasing win?
# If team batted first and won by runs -> Defended (Bat 1st)
# If team won by wickets -> Chased (Bat 2nd)
def match_result_type(row):
    if row['margin_type'] == 'Runs':
        return 'Defended (Bat 1st Won)'
    elif row['margin_type'] == 'Wickets':
        return 'Chased (Bat 2nd Won)'
    return 'Other'

df['result_type'] = df.apply(match_result_type, axis=1)

print("\nEngineered Data Sample:")
print(df[['venue', 'winner', 'result_type', 'margin_value', 'margin_type', 'top_scorer_pct_contribution']].head())

# ==========================================
# 4. EXPLORATORY DATA ANALYSIS (EDA)
# ==========================================

# Chart 1: Toss Decision & Match Result (Bat 1st vs Chasing)
fig, ax = plt.subplots(1, 2, figsize=(14, 5))

df['result_type'].value_counts().plot(
    kind='pie', autopct='%1.1f%%', ax=ax[0], startangle=90, colors=['#4CAF50', '#2196F3', '#FF9800']
)
ax[0].set_title('Match Outcome: Defended vs Chased', fontsize=12, fontweight='bold')
ax[0].set_ylabel('')

df['winner'].value_counts().plot(kind='bar', ax=ax[1], color='steelblue')
ax[1].set_title('Total Wins per Team in T20 WC 2022', fontsize=12, fontweight='bold')
ax[1].set_xlabel('Team')
ax[1].set_ylabel('Matches Won')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()

# Chart 2: Top 10 Run Contributors & Top Scorers
fig, ax = plt.subplots(1, 2, figsize=(15, 5))

top_scorers = df.groupby('top scorer')['highest score'].sum().sort_values(ascending=False).head(10)
top_scorers.plot(kind='bar', ax=ax[0], color='coral')
ax[0].set_title('Top 10 Run Aggregators (Sum of Top Scores)', fontsize=12, fontweight='bold')
ax[0].set_ylabel('Runs')
ax[0].tick_params(axis='x', rotation=45)

top_contribution = df.groupby('winner')['top_scorer_pct_contribution'].mean().sort_values(ascending=False).head(10)
top_contribution.plot(kind='bar', ax=ax[1], color='darkcyan')
ax[1].set_title('Avg % Team Score from Top Scorer (Per Team)', fontsize=12, fontweight='bold')
ax[1].set_ylabel('% Contribution')
ax[1].tick_params(axis='x', rotation=45)

plt.tight_layout()
plt.show()

# Chart 3: Venue Profiling (Average 1st Innings Par Score)
plt.figure(figsize=(10, 5))
venue_stats = df.groupby('venue')['first innings score'].mean().sort_values(ascending=False)
sns.barplot(x=venue_stats.values, y=venue_stats.index, palette='crest')
plt.title('Average 1st Innings Par Score by Venue', fontsize=13, fontweight='bold')
plt.xlabel('Average First Innings Score')
plt.ylabel('Venue')
plt.tight_layout()
plt.show()

# ==========================================
# 5. MACHINE LEARNING PIPELINE
# ==========================================

print("\n" + "="*50)
print("   MACHINE LEARNING: MATCH WINNER PREDICTION   ")
print("="*50)

# Goal: Predict if Team 1 will win the match
# Features available: venue, team1, team2, toss winner, toss decision, 1st innings score

ml_df = df[['venue', 'team1', 'team2', 'toss winner', 'toss decision', 'first innings score', 'winner']].dropna().copy()

# Binary Target: 1 if team1 won, 0 if team2 won
ml_df['team1_won'] = (ml_df['winner'] == ml_df['team1']).astype(int)

# Categorical feature encoding
feature_cols = ['venue', 'team1', 'team2', 'toss winner', 'toss decision']
label_encoders = {}

for col in feature_cols:
    le = LabelEncoder()
    # Fit encoder on all unique team/venue values
    ml_df[col] = le.fit_transform(ml_df[col].astype(str))
    label_encoders[col] = le

# Define X and y
X = ml_df[feature_cols + ['first innings score']]
y = ml_df['team1_won']

# Train/Test Split (Stratified to maintain 0/1 balance)
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=42, stratify=y
)

# ----------------------------------------------------
# Model A: Logistic Regression (Prevents overfitting on small datasets)
# ----------------------------------------------------
lr = LogisticRegression(max_iter=1000, random_state=42)
lr.fit(X_train, y_train)
y_pred_lr = lr.predict(X_test)

print(f"\n[Logistic Regression] Test Accuracy: {accuracy_score(y_test, y_pred_lr):.2%}")

# ----------------------------------------------------
# Model B: Random Forest Classifier (Captures non-linear interactions)
# ----------------------------------------------------
rf = RandomForestClassifier(n_estimators=150, max_depth=4, random_state=42)
rf.fit(X_train, y_train)
y_pred_rf = rf.predict(X_test)

print(f"[Random Forest]       Test Accuracy: {accuracy_score(y_test, y_pred_rf):.2%}")
print("\nClassification Report (Random Forest):\n", classification_report(y_test, y_pred_rf))

# Confusion Matrix
plt.figure(figsize=(5, 4))
sns.heatmap(confusion_matrix(y_test, y_pred_rf), annot=True, fmt='d', cmap='Blues',
            xticklabels=['Team 2 Won', 'Team 1 Won'], yticklabels=['Team 2 Won', 'Team 1 Won'])
plt.title('Random Forest - Confusion Matrix')
plt.xlabel('Predicted')
plt.ylabel('Actual')
plt.tight_layout()
plt.show()

# Feature Importance
feature_importance = pd.Series(rf.feature_importances_, index=X.columns).sort_values(ascending=True)
plt.figure(figsize=(8, 4))
feature_importance.plot(kind='barh', color='darkseagreen')
plt.title('Feature Importance for Match Winner Prediction', fontsize=12, fontweight='bold')
plt.xlabel('Importance Score')
plt.tight_layout()
plt.show()

# ==========================================
# 6. INTERACTIVE MATCH PREDICTOR FUNCTION
def predict_custom_match(venue, team1, team2, toss_winner, toss_decision, first_innings_score):
    """
    Predicts the winning probability for any matchup scenario.
    """
    try:
        encoded_input = [
            label_encoders['venue'].transform([venue])[0],
            label_encoders['team1'].transform([team1])[0],
            label_encoders['team2'].transform([team2])[0],
            label_encoders['toss winner'].transform([toss_winner])[0],
            label_encoders['toss decision'].transform([toss_decision])[0],
            first_innings_score
        ]
        prob = rf.predict_proba([encoded_input])[0]
        print(f"\n--- Match Prediction at {venue} ---")
        print(f"Target Set by {team1}: {first_innings_score} runs")
        print(f"{team1} Win Probability: {prob[1]:.1%}")
        print(f"{team2} Win Probability: {prob[0]:.1%}")
        predicted_winner = team1 if prob[1] > prob[0] else team2
        print(f"Predicted Winner: {predicted_winner}")
    except ValueError as e:
        print(f"Error: One of the team/venue names was not in the 2022 dataset: {e}")

# Example Simulation: India vs Pakistan at Melbourne
# (Only call if teams exist in dataset categories)
available_teams = list(df['team1'].unique())
available_venues = list(df['venue'].unique())

if 'India' in available_teams and 'Pakistan' in available_teams and len(available_venues) > 0:
    predict_custom_match(
        venue=available_venues[0],
        team1='India',
        team2='Pakistan',
        toss_winner='Pakistan',
        toss_decision='field',
        first_innings_score=160
    )
```
